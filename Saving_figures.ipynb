```python
import matplotlib.pyplot as plt
import numpy as np

from pathlib import Path

BASE_DIR = Path.cwd()

%matplotlib qt
```

```python
def read_data(filename):
    return np.loadtxt(
        filename,
        delimiter="\t",skiprows=1,
        converters=lambda x: float(x.replace(",", ".")))
```

```python
#c_data_1 = read_data("txtfiles/DATA_COMPR_CUT.tab")
c_data_1 = read_data("txtfiles/COMPRAVG_DS_3.tab")
c_data_2 = read_data("txtfiles/COMPRAVG_DS_2.tab")

#t_data_1 = read_data("txtfiles/DATA_TENS_CUT.tab")
t_data_1 = read_data("txtfiles/TENSAVG_DS_3.tab")
t_data_2 = read_data("txtfiles/TENSAVG_DS_2.tab")

#ct_data_1 = read_data("txtfiles/DATA_AVG_CONC.tab")
ct_data_1 = read_data("txtfiles/DATA_AVG_DS_CONC2.tab")
ct_data_2 = read_data("txtfiles/DATA_AVG_DS_CONC.tab")
```

```python
lam1 = c_data_1[:,0]
sigma1 = c_data_1[:,1]/10**6

lam2 = c_data_2[:,0]
sigma2 = c_data_2[:,1]/10**6

lam3 = t_data_1[:,0]
sigma3 = t_data_1[:,1]/10**6

lam4 = t_data_2[:,0]
sigma4 = t_data_2[:,1]/10**6

lam5 = ct_data_1[:,0]
sigma5 = ct_data_1[:,1]/10**6

lam6 = ct_data_2[:,0]
sigma6 = ct_data_2[:,1]/10**6

```

```python


fig,ax = plt.subplots()

ax.grid()
ax.legend()
ax.set_title("Nyomóvizsgálat")
ax.set_xlabel(r"$\lambda$, 1")
ax.set_ylabel(r"$P$, MPa")

#ax.scatter(lam1, sigma1, s=50, zorder=2)
ax.plot(lam1, sigma1, linewidth=4)
```

```python
fig,ax = plt.subplots()

ax.grid()
ax.legend()

ax.scatter(lam2, sigma2, s=50, zorder=2)
ax.plot(lam1, sigma1, alpha=0.5, linewidth=3, zorder=1)
```
