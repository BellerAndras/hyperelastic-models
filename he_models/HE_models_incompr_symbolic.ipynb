```python
import sympy as sp
```

Neo-Hooke, inkompr, UNIAX

```python
lam1, lam2, lam3, lamT, C10 , C11,p, sigma1, sigma2, sigma3 = sp.symbols(
    'lambda_1 lambda_2 lambda_3 lambda_T C_10 C_11 p sigma_1 sigma_2 sigma_3')

W_NH = C10*(lam1**2+lam2**2+lam3**2-3)

eq1 = sp.Eq(sigma1,lam1*sp.diff(W_NH,lam1)+p).subs(lam2,lam1**(-1/2))
eq2 = sp.Eq(sigma2, lam2*sp.diff(W_NH,lam2)+p).subs(lam2,lam1**(-1/2))
eq3 = sp.Eq(sigma2,0)

sol1 = sp.solve([eq1,eq2,eq3],[p,sigma1,sigma2])
sol1 = list(sol1.values())

display(sol1[0])
display(sol1[1])
```
