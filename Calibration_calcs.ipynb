```python
#input number is the syringe's position, result in MPa

def syr_pos_to_pa(syr_pos,p0=0.1,Vin=10000):
    V_syr_disp_ml = 100-syr_pos #displaced volume, ml
    Vsyr_mm3 = 1000*V_syr_disp_ml #displaced volume, mm3
    p2_abs = p0*(Vin+Vsyr_mm3)/Vin
    p2_gauge = p2_abs-p0
    print(p2_gauge)
    return p2_gauge

syr_pos_to_pa(38)
```
