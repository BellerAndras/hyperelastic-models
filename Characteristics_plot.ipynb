```python
import matplotlib.pyplot as plt
import pandas as pd
```

```python


df = pd.read_csv(
    "plot1.rpt",
    sep=r"\s+",
    header=None,
    engine="python"
)

arr = df.iloc[1:].astype(float).to_numpy().T

fig, ax1 = plt.subplots()

ax1.grid()
ax1.set_aspect("equal")
ax1.plot(arr[0], arr[1])

plt.show()
```
